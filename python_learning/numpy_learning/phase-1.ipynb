## Numpy Array & Basics

Learning basics of numpy in `phase-1`

```python
import numpy as np
```

### Creating array from list

```python
arr_1d = np.array([1,2,3,4,5])
print("1D array: \n", arr_1d)

arr_2d = np.array([[1,2,3], [4,5,6]])
print("2D array: \n", arr_2d)
```

### List VS Numpy array

```python
py_list = [1,2,3]
print("Python list multiplication : \n", py_list * 2)

np_array = np.array([1,2,3]) # element wise multiplication
print("Python array(numpy) multiplication : \n", np_array * 2)

import time
start = time.time()
py_list = [i*2 for i in range(10000000)]
print("\n List operation time: \n", time.time() - start)

start = time.time()
np_array = np.arange(10000000) * 2
print("\n Numpy operation time: \n", time.time() - start)
```

### Creating array from scratch

```python
zeros = np.zeros((3,4))
print("zeros array: \n", zeros)

ones = np.ones((2,3))
print("ones array: \n", ones)

full = np.full((2,2), 7)
print("full array: \n", full)

random = np.random.random((2,3))
print("random array: \n", random)

sequence = np.arange(0,11,2)
print("sequence array: \n", sequence)
```

### Vector, Matrices & Tensor

```python
vector = np.array([1,2,3])
print("Vector: \n", vector)

matrices = np.array([[1,2,3],
                   [4,5,6]])
print("Matrices: \n", matrices)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6], [7,8]]])
print("Matrices: \n", tensor)
```

### Array Properties

```python
arr = np.array([[1,2,3],
                [4,5,6]])
print("Shape: ", arr.shape)
print("Dimension: ", arr.ndim)
print("Size: ", arr.size)
print("DType: ", arr.dtype)
```

### Array Reshaping

```python
arr = np.arange(12)
print("Original Array: ", arr)

reshaped = arr.reshape((3,4))
print("Reshapes Array: \n", reshaped)

flattened = reshaped.flatten()
print("Flattened Array: \n", flattened)

# ravel (returns a view instead of copy)
raveled = reshaped.ravel()
print("Raveled Array: \n", raveled)

transpose = reshaped.T
print("Transposed Array: \n", transpose)
```
