### Exercise 2: Convert 1D Array to 2D

Write a code snippet to convert a 1D array to a 2D array with 2 rows.

**Given:**
```python
import numpy as np
arr = np.arange(6)
```

**Expected Output:**
```text
Original Array: [0 1 2 3 4 5] 
Reshaped 2x3 Array: 
[[0 1 2]
 [3 4 5]]
```


````python
orig_arr = np.arange(6)
print('Original Array: ', orig_arr)

reshaped_arr = orig_arr.reshape(2,3)
print('Reshaped Array: \n', reshaped_arr)
````

### Exercise 3: Print Array Attributes

**Instructions**
Print the following attributes of the array:
* The shape of the array.
* The number of array dimensions.
* The size of each element in bytes.

**Given:**
```python
import numpy as np
my_array = np.array([[1, 2], [3, 4], [5, 6], [7, 8]], dtype=np.uint16)
```

**Expected Output:**
```text
The shape of the array is: (4, 2)
The number of dimensions is: 2
The size of each element in bytes is: 2
```

````python
my_array = np.array([[1, 2], [3, 4], [5, 6], [7, 8]], dtype=np.uint16)
print('The shape of the array is: ', my_array.shape)
print('The number of dimensions is: ', my_array.ndim)
print('The size of each element in bytes is: ', my_array.itemsize)
````

````python
sampleArray = np.array([[34, 43, 73], [82, 22, 12], [53, 94, 66]])

# Get the indices that would sort the array by its second column
sorted_indices = sampleArray[:, 1].argsort()

# Use the sorted indices to reorder the rows of the array
sorted_array = sampleArray[sorted_indices]

print("Original array: \n", sampleArray)
print("Sorted array: \n", sorted_array)
````

````python
sampleArray = np.array([[34,43,73],[82,22,12],[53,94,66]])
newColumnToAdd = np.array([10, 10, 10])

# Delete the second column (index 1)
# axis=1 specifies that the operation should be on columns
deletedArray = np.delete(sampleArray, 1, axis=1)

# Insert the new column at the second position (index 1)
# axis=1 specifies that we're inserting a column
resultArray = np.insert(deletedArray, 1, newColumnToAdd, axis=1)

print(resultArray)
````

````python
# 1. Box A: The coefficients (numbers in front of x and y)
A = np.array([[1, 2],
              [3, 4]])

# 2. Box B: The constant answers on the right side
B = np.array([8, 18])

# 3. Let NumPy do the heavy lifting
solution = np.linalg.solve(A, B)

print("Value of x:", solution[0])
print("Value of y:", solution[1])
````

````python
arr = np.array([1.5, 2.8, 3.2, 4.1])
target_value = 3

# 1. Find the index of the closest number to 3
closest_idx = np.abs(arr-target_value).argmin()

# 2. Extract the actual value using that index
closest_value = arr[closest_idx]

print(f"The value closest to {target_value} is: {closest_value}")
````

````python
arr = np.array([1, 2, 3, 4, 5])

# Create an empty object array with the same size
obj_arr = np.empty(arr.size, dtype='object')

# Assign the elements from the original array
obj_arr[:] = arr

# We can now add a string element to it
obj_arr[1] = 'a'
print(obj_arr)
print(obj_arr.dtype)
````
